```python
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score, confusion_matrix, classification_report
import matplotlib.pyplot as plt
import numpy as np

# 1. Prepare binary target
le = LabelEncoder()
y_binary = le.fit_transform(wine_quality['quality_category'])
# The LabelEncoder will map 'Bad' to 0 and 'Good' to 1 (alphabetical order)
print(f"Original classes and their encoded values: {list(zip(le.classes_, le.transform(le.classes_)))}")

# 2. Data Splitting for Binary Classification
# X was already defined in a previous cell by dropping 'quality' and 'quality_category'
# We need to re-split X and y_binary to ensure correct train/test sets for binary classification
X_train_binary, X_test_binary, y_train_binary, y_test_binary = train_test_split(
    X, y_binary, test_size=0.2, random_state=42, stratify=y_binary
)

# 3. Feature Scaling
scaler_binary = StandardScaler()
X_train_binary_scaled = scaler_binary.fit_transform(X_train_binary)
X_test_binary_scaled = scaler_binary.transform(X_test_binary)

# 4. Train a Binary Classifier (Logistic Regression)
model_binary = LogisticRegression(max_iter=5000, solver='liblinear', random_state=42)
model_binary.fit(X_train_binary_scaled, y_train_binary)

# 5. Predict Probabilities
y_pred_proba = model_binary.predict_proba(X_test_binary_scaled)[:, 1] # Probability of the positive class (Good)

# 6. Threshold Tuning Loop
thresholds = np.arange(0.1, 1.0, 0.01) # Test thresholds from 0.1 to 0.99
accuracies = []
precisions = []
recalls = []
f1_scores = []

for t in thresholds:
    y_pred_tuned = (y_pred_proba >= t).astype(int)
    accuracies.append(accuracy_score(y_test_binary, y_pred_tuned))
    precisions.append(precision_score(y_test_binary, y_pred_tuned, zero_division=0)) # Positive class is 1 (Good)
    recalls.append(recall_score(y_test_binary, y_pred_tuned, zero_division=0))    # Positive class is 1 (Good)
    f1_scores.append(f1_score(y_test_binary, y_pred_tuned, zero_division=0))       # Positive class is 1 (Good)

# 7. Identify Optimal Threshold (e.g., based on F1-score for the positive class 'Good')
best_f1_idx = np.argmax(f1_scores)
optimal_threshold = thresholds[best_f1_idx]

print(f"\nOptimal Threshold (based on F1-score for 'Good' wine): {optimal_threshold:.2f}")

# 8. Evaluate with Optimal Threshold
y_pred_optimal = (y_pred_proba >= optimal_threshold).astype(int)
print("\n--- Metrics with Optimal Threshold ---")
print(f"Accuracy: {accuracy_score(y_test_binary, y_pred_optimal):.4f}")
print(f"Precision (Good): {precision_score(y_test_binary, y_pred_optimal, zero_division=0):.4f}")
print(f"Recall (Good): {recall_score(y_test_binary, y_pred_optimal, zero_division=0):.4f}")
print(f"F1-Score (Good): {f1_score(y_test_binary, y_pred_optimal, zero_division=0):.4f}")
print("\nConfusion Matrix:\n", confusion_matrix(y_test_binary, y_pred_optimal))
print("\nClassification Report:\n", classification_report(y_test_binary, y_pred_optimal, target_names=le.classes_, zero_division=0))

# 9. Visualization
plt.figure(figsize=(12, 7))
plt.plot(thresholds, accuracies, label='Accuracy', color='blue')
plt.plot(thresholds, precisions, label='Precision (Good)', color='green')
plt.plot(thresholds, recalls, label='Recall (Good)', color='red')
plt.plot(thresholds, f1_scores, label='F1-Score (Good)', color='purple', linewidth=2)
plt.xlabel('Probability Threshold')
plt.ylabel('Score')
plt.title('Performance Metrics vs. Probability Threshold (Logistic Regression)')
plt.axvline(x=optimal_threshold, color='orange', linestyle='--', label=f'Optimal Threshold (F1-score): {optimal_threshold:.2f}')
plt.legend()
plt.grid(True)
plt.ylim(0, 1.05)
plt.show()
```
